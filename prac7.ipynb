```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
forest = pd.read_csv("forestfires.csv")

print(forest.head())
```

```python
print(forest.info())
```

```python
forest.hist(figsize=(12,10))

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(forest['temp'], forest['area'])

plt.xlabel("Temperature")

plt.ylabel("Burned Area")

plt.title("Temperature vs Burned Area")

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.plot(forest['temp'])

plt.title("Temperature Variation")

plt.xlabel("Index")

plt.ylabel("Temperature")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.barplot(x='month', y='temp', data=forest)

plt.title("Average Temperature by Month")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.countplot(x='month', data=forest)

plt.title("Forest Fires by Month")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(y='temp', data=forest)

plt.title("Temperature Distribution")

plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(forest.corr(numeric_only=True), annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")

plt.show()
```

```python
sns.pairplot(forest)

plt.show()
```
